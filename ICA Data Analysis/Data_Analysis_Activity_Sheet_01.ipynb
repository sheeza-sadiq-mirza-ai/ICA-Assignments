Data set 01

```python
e = pd.read_csv(r"Downloads\pakistani_ecommerce_dataset.csv")
```

Most popular product category

```python
e['product_category'].mode()
```

Average order value 

```python
e['order_value'].mean().round(2)
```

City with the highest number of orders

```python
e['city'].mode()
```

Most used payment method 

```python
e['payment_method'].mode()
```

Bar chart showing average order value by city

```python
e.groupby('city')['order_value'].agg(['mean']).round(2)
```

```python
categories = ['Faisalabad', 'Islamabad', 'Karachi', 'Lahore', 'Rawalpindi']
values = [2937.08, 3011.95, 2956.50, 2785.33, 2447.08]
plt.bar(categories, values, color='darkseagreen', width=0.5)
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Bar Plot')
plt.show()
```

Data set 02 

```python
p = pd.read_csv(r"Downloads\student_lifestyle_dataset.csv")
```

```python
p
```

Average number of study hours

```python
p['study_hours'].mean().round(2)
```

Who sleeps more on average

```python
p.groupby('gender')['sleep_hours'].agg(['mean']).round(2)
```

Relationship between screentime and sleep 

```python
p['screen_time'].corr(p['sleep_hours']).round(2)
```

```python
# very weak positive relationship 
```

Histogram of daily transport cost

```python
plt.hist(p['daily_transport_cost'], bins=20, color='powderblue', edgecolor='black', alpha=0.7)
plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.title('Histogram')
```
